```python
import cv2
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
```

```python
# Load image using OpenCV (automatically stored as a NumPy array)
img = cv2.imread('Images/cameraman.tif')

# Display the image using Matplotlib
plt.imshow(img)
plt.title("Loaded via OpenCV")
plt.axis('off')  # Hide axis coordinates
plt.show()
```

```python
# Load image using PIL (stored as a PIL Image object)
img2 = Image.open('Images/lena_gray_256.tif')

# Display the image with a grayscale colormap
plt.imshow(img2, cmap=cm.Greys_r)
plt.title("Loaded via PIL (Grayscale)")
plt.axis('off')
plt.show()
```

```python
# Save the OpenCV image to disk
cv2.imwrite('new_image.jpg', img)

# Save the PIL image to disk
img2.save('new_image2.jpg')

print("Both images have been saved successfully.")
```

```python
# Check shape and print pixel values for the OpenCV image
print("=== OpenCV Image ===")
print("Dimensions (Shape):", img.shape)
print("Pixel Array:\n", img)

# Convert PIL image to a NumPy array, then inspect shape and values
img_array = np.array(img2)
print("\n=== PIL Image (Converted to NumPy) ===")
print("Dimensions (Shape):", img_array.shape)
print("Pixel Array:\n", img_array)
```

```python
# Inspect data type, minimum, and maximum pixel values
print("Data Type:", img_array.dtype)
print("Min Pixel Value:", np.min(img_array))
print("Max Pixel Value:", np.max(img_array))

# Crop a region of interest using array slicing
cropped_img = img_array[50:200, 50:200]

# Visualize the cropped region
plt.imshow(cropped_img, cmap='gray')
plt.title("Cropped Region")
plt.axis('off')
plt.show()
```
