```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
```

```python
# Load a sample image
image = data.astronaut()

# Save the image
cv2.imwrite("Parrot.png", cv2.cvtColor(image, cv2.COLOR_RGB2BGR))

# Convert to float32
image_float = image.astype(np.float32) / 255.0

plt.figure(figsize=(6, 5))
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
# Apply a 7x7 box filter
smoothed = cv2.blur(image, (7, 7))

# Show original and smoothed images
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(smoothed)
ax[1].set_title("7x7 Box Filter")
ax[1].axis("off")

plt.show()
```

```python
# Apply 5x5 Gaussian filter
gaussian_5 = cv2.GaussianBlur(image, (5, 5), 0)

# Apply 21x21 Gaussian filter
gaussian_21 = cv2.GaussianBlur(image, (21, 21), 0)

# Show the three images
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(gaussian_5)
ax[1].set_title("5x5 Gaussian Filter")
ax[1].axis("off")

ax[2].imshow(gaussian_21)
ax[2].set_title("21x21 Gaussian Filter")
ax[2].axis("off")

plt.show()
```

```python
# Convert image to float32
image_float = image.astype(np.float32) / 255.0

# Apply a 3x3 Laplacian filter
laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=3)

# Sharpen the image
sharpened = np.clip(image_float - laplacian, 0, 1)

# Show the results
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].imshow(image_float)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(laplacian, cmap="gray")
ax[1].set_title("Laplacian Filter")
ax[1].axis("off")

ax[2].imshow(sharpened)
ax[2].set_title("Sharpened Image")
ax[2].axis("off")

plt.show()
```
