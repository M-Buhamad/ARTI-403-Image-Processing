```python
import matplotlib.pyplot as plt
import numpy as np
from skimage import data, exposure
```

```python
# Task 1: Rescale intensity values between 3rd and 80th percentiles
moon_img = data.moon()

p3, p80 = np.percentile(moon_img, (3, 80))
img_rescaled = exposure.rescale_intensity(moon_img, in_range=(p3, p80))

# Plot result and histogram
fig, axes = plt.subplots(2, 1, figsize=(8, 6))

axes[0].imshow(img_rescaled, cmap='gray')
axes[0].set_title('Rescaled Image (3rd - 80th Percentile)')
axes[0].axis('off')

axes[1].hist(img_rescaled.ravel(), bins=256, range=(0, 255))
axes[1].set_title('Histogram')
axes[1].set_xlabel('Gray levels')
axes[1].set_ylabel('Pixel count')

plt.tight_layout()
plt.show()
```

```python
# Task 2: Equalize histogram using exposure.equalize_hist
moon_img = data.moon()

img_eq = exposure.equalize_hist(moon_img)

# Plot equalized image and its histogram
fig, axes = plt.subplots(2, 1, figsize=(8, 6))

axes[0].imshow(img_eq, cmap='gray')
axes[0].set_title('Equalized Image')
axes[0].axis('off')

axes[1].hist(img_eq.ravel(), bins=256)
axes[1].set_title('Equalized Histogram')
axes[1].set_xlabel('Intensity values')
axes[1].set_ylabel('Pixel count')

plt.tight_layout()
plt.show()
```

```python
# Task 3: Match histogram of Chelsea image to Rocket image
reference_img = data.rocket()
image_to_match = data.chelsea()

matched_img = exposure.match_histograms(image_to_match, reference_img, channel_axis=-1)

# Display source, reference, and matched images
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

axes[0].imshow(image_to_match)
axes[0].set_title('Source (Chelsea)')
axes[0].axis('off')

axes[1].imshow(reference_img)
axes[1].set_title('Reference (Rocket)')
axes[1].axis('off')

axes[2].imshow(matched_img)
axes[2].set_title('Matched Result')
axes[2].axis('off')

plt.tight_layout()
plt.show()
```
